# PyMC Model TrueMeasure in QMCPy

[PyMC](https://www.pymc.io/), a popular Probabilistic Programming Language (PPL), allows specifiying statistical models in an easy to use and familiar syntax. A PyMC `Model` combines several probability distributions and other operations into one joint distribution of random variables. `PyMCWrapper` allows using a PyMC `Model` as a QMCPy TrueMeasure.

This tutorial shows basic examples of constructing and computing expectations with PyMC models, and covers limitations and workarounds around these limitations.

We state limitations an gotchas here so the reader can easily find them below:
- The samples correspond strictly to the random variables, **not** to any transformation of them or any Deterministics in the model.
- Multivariate distributions are not supported (see workaround below).
- Only sampling from the prior distribution is supported. Posterior sampling takes extra steps shown below.
- Discrete and some other distributions are currently not supported as their icdf is not currently defined by PyMC.

## 1. Basic Use

For example, consider
$$
X\sim \,N(0.0,1.0).
$$

````python
with pm.Model() as model:
    x = pm.Normal("x", mu=0.0, sigma=1.0)

tm_normal = PyMCWrapper(DigitalNetB2(1, seed=7), model)
samples = tm_normal(2**12)
````

````python
plt.hist(samples, density=True, bins=30);
plt.title("PyMC Normal Model Samples");
plt.xlabel("x");
plt.ylabel("density");
````

As we can see, the `PyMCWrapper` is initialized with
```
PyMCWrapper(discrete_distribution, model)
```
where the `discrete_distribution` must have the right number of dimensions, according to the distributions in the pymc model and their shapes. If we use a distribution with the wrong dimension we get **an error telling us the right dimension to use**:

````python
try:
    tm_normal = PyMCWrapper(DigitalNetB2(2, seed=7), model) # intentional error, shown below
except DimensionError as error:
    print(f"Caught Error: [\033[1m{str(error)}]")
````

The error message is telling us to use a sampler with 1 dimension for this model.  
We can also get weights (1/pdf) values for the samples by setting `get_weights=True`:

````python
nu = 10.0
with pm.Model() as chisquare:
    x = pm.ChiSquared("s", nu=nu)

tm_chisquare = PyMCWrapper(DigitalNetB2(1, seed=7), chisquare)
samples, weights = tm_chisquare(2**12, return_weights=True)
plt.hist(samples, density=True, bins=100, label="samples");
plt.scatter(samples, 1/weights, color="green", label="1/weight", alpha=0.3, linewidth=1);
xs = np.linspace(1e-6,35,1000)
pdf = stats.chi2.pdf(xs, df=nu)
plt.plot(xs, pdf, label="True PDF", color="black", alpha=0.8, linewidth=2)
plt.title("PyMC ChiSquared Model Samples and pdf");
plt.xlabel("x");
plt.ylabel("density");
plt.legend();
````

## 2. Hierarchical Specification of Student's t-distribution

Here we show a more realistic use of this feature and compare results with MCMC.  
A Student's t-distributed $X$ can be written as
$$
\begin{align*}
s \sim & \chi^2(\nu)\\
X \sim & N(0,\sqrt{\nu/s})
\end{align*}
$$

We show this hierarchical construction gives the same results as the marginalized one by comparing samples.

````python
nu = 5.0
n_replications=2**7
with pm.Model() as tstudent_normal_model:
  s = pm.ChiSquared("s", nu=nu)
  x = pm.Normal("x", mu=0.0, sigma=pm.math.sqrt(nu/s))

# take samples using MCMC
with tstudent_normal_model:
  idata = pm.sample(draws=2048*n_replications, tune=1000)
mcmc_s_samples = np.asarray(idata.posterior["s"]).ravel()
mcmc_x_samples = np.asarray(idata.posterior["x"]).ravel()
````

````python
# take samples using QMC
tstudent_normal = PyMCWrapper(DigitalNetB2(2, replications=n_replications, seed=7), tstudent_normal_model)
tstudent_normal_iid = PyMCWrapper(IIDStdUniform(2, replications=n_replications, seed=7), tstudent_normal_model)
qmc_samples = tstudent_normal(2**15)
qmc_s_samples = qmc_samples[...,0]
qmc_x_samples = qmc_samples[...,1]
iid_samples = tstudent_normal_iid(2**15)
iid_x_samples = iid_samples[...,1]
````

````python
fig, ax = plt.subplots(1, 2, figsize=(12, 8))
ax[0].scatter(np.sqrt(nu/qmc_s_samples[0,:]), qmc_x_samples[0,:], alpha=0.1, label="qmc samples");
ax[0].scatter(np.sqrt(nu/mcmc_s_samples[::n_replications]), mcmc_x_samples[::n_replications], label="mcmc samples", alpha=0.1)
ax[0].legend();
ax[0].set_xlabel(r"$\sigma=\sqrt{\nu/s}$");
ax[0].set_ylabel("x");
ax[0].set_xlim(0.4,3);
ax[0].set_ylim(-4,4);

xs = np.linspace(-12,12,1000)
pdf = stats.t.pdf(xs, nu)
ax[1].hist(qmc_x_samples[0,:], bins=100, alpha=0.5, label="qmc samples", density=True);
ax[1].hist(mcmc_x_samples[::n_replications], bins=100, label="mcmc samples", alpha=0.5, density=True)
ax[1].plot(xs, pdf, label="True PDF", color="black")
ax[1].legend();
ax[1].set_xlabel("x");
ax[1].set_yscale("log")
````

As shown in the plots, the quasi Monte Carlo samples are more uniformly spread out, with the Markov Chain Monte Carlo samples being more concentrated around the mode and tend to cluster together. To verify the results and compare with MCMC we can compute known statistics of this distribution and compare:

````python
qmc_results = describe(qmc_x_samples.ravel())
mcmc_results = describe(mcmc_x_samples)
iid_results = describe(iid_x_samples.ravel())
kurtosis = np.inf if (2< nu <= 4) else (6/(nu-4))
exact_results = {"mean": 0.0, "variance": nu/(nu-2), "skewness": 0.0, "kurtosis": kurtosis}
rows = ["mean", "variance", "skewness", "kurtosis"]
show_table(
    ["Type"]+ ["#samples"]+[r.capitalize() for r in rows],
    [
    ["Exact"]+["NA"]+[exact_results[r] for r in rows],
    ["MCMC"]+[mcmc_results.nobs]+[getattr(mcmc_results,r) for r in rows],
    ["QMC"]+[qmc_results.nobs]+[getattr(qmc_results, r) for r in rows],
    ["IID"]+[iid_results.nobs]+[getattr(iid_results, r) for r in rows]
    ]
)
````

We can see that estimation with QMC is at least as precise at the same number of samples.  We can also compare the behavior of the error for each of these estimates for the QMC, MCMC and IID inverse transform sampling cases:

````python
fig, axs = plt.subplots(2,2, layout="constrained")
axs = axs.ravel()
for ax, name, op in zip(axs, ["mean", "variance", "skewness", "kurtosis"], [np.mean, np.var, stats.skew, stats.kurtosis]):
    exact = exact_results[name]
    samples_num = 2**np.arange(3,16)
    mcmc_res = np.array([op(mcmc_x_samples.reshape(n_replications, -1)[:,:n], axis=1) for n in samples_num])
    qmc_res = np.array([op(qmc_x_samples[:,:n], axis=1) for n in samples_num])
    iid_res = np.array([op(iid_x_samples[:,:n], axis=1) for n in samples_num])
    mcmc_error = np.abs((mcmc_res - exact)/np.abs(exact))*100 if name not in ["mean", "skewness"] else np.abs(mcmc_res - exact)
    qmc_error = np.abs((qmc_res - exact)/np.abs(exact))*100 if name not in ["mean", "skewness"] else np.abs(qmc_res - exact)
    iid_error = np.abs((iid_res - exact)/np.abs(exact))*100 if name not in ["mean", "skewness"] else np.abs(iid_res - exact)
    ax.plot(samples_num, np.median(mcmc_error, axis=1), label="MCMC error", color="blue")
    ax.fill_between(samples_num, np.quantile(mcmc_error, 0.1, axis=1), np.quantile(mcmc_error, 0.9, axis=1), color="blue", alpha=0.1)
    ax.plot(samples_num, np.median(qmc_error, axis=1), label="QMC error", color="orange")
    ax.fill_between(samples_num, np.quantile(qmc_error, 0.1, axis=1), np.quantile(qmc_error, 0.9, axis=1), color="orange", alpha=0.1)
    ax.plot(samples_num, np.median(iid_error, axis=1), label="IID error", color="green")
    ax.fill_between(samples_num, np.quantile(iid_error, 0.1, axis=1), np.quantile(iid_error, 0.9, axis=1), color="green", alpha=0.1)
    #if name in ["mean", "skewness", "variance"]:
    ax.set_yscale("log")
    ax.set_xscale("log")
    ax.set_title(name+ (r" \%error" if name not in ["mean", "skewness"] else " error"))
plt.legend();
````

## 3. Sampling In Constrained vs Unconstrained Space

Many random variables are defined on a subspace of $R^n$, such as the case for the $\chi^2(k)$ distribution. It is sometimes useful to transform these variables in a way that renders them "unconstrained" so they are defined on $R^n$. This is done internally by PyMC during MCMC sampling and other fitting procedures.  

Here we demonstrate obtaining samples and weights in the unconstrained space. The unconstrained space is defined by the PyMC model, where if no transformation is given by the user, a `default_transform` is used by PyMC which is distribution dependent.  
The samples in the unconstrained space can be obtained by setting the `unconstrained_samples` option to `True`:

````python
nu = 10.0
tm_chisquare = PyMCWrapper(DigitalNetB2(1, seed=7), chisquare)
tm_unconstrained_chisquare = PyMCWrapper(DigitalNetB2(1, seed=7), chisquare, unconstrained_samples=True)

fig, ax = plt.subplots(1,3, layout="constrained")
samples              , weights               = tm_chisquare(2**12,               return_weights=True)
unconstrained_samples, unconstrained_weights = tm_unconstrained_chisquare(2**12, return_weights=True)
ys = np.linspace(1e-6,35,1000)
pdf = stats.chi2.pdf(ys, df=nu)
xs = np.linspace(1e-6,4,1000)
pdf_unconstrained = stats.chi2.pdf(np.exp(xs), df=nu) * np.exp(xs)

ax[0].hist(samples, density=True, bins=100, label="samples");
ax[0].plot(ys, pdf, label="True PDF", color="black", alpha=0.8, linewidth=2)
ax[0].scatter(samples, 1/weights, color="green", label="1/weight", alpha=0.3, linewidth=1);
ax[0].set_title("constrained_samples")

ax[1].hist(unconstrained_samples, density=True, bins=100, label="samples");
ax[1].plot(xs, pdf_unconstrained, label="True PDF", color="black", alpha=0.8, linewidth=2)
ax[1].scatter(unconstrained_samples, 1/unconstrained_weights, color="green", label="1/weight", alpha=0.3, linewidth=1);
ax[1].set_title("unconstrained samples")

ax[2].hist(np.exp(unconstrained_samples), density=True, bins=100, label="samples");
ax[2].plot(ys, pdf, label="True PDF", color="black", alpha=0.8, linewidth=2)
ax[2].scatter(samples, 1/weights, color="green", label="1/weight", alpha=0.3, linewidth=1);
ax[2].set_title("transformed unconstrained samples")

ax[0].set_xlabel("x");
ax[1].set_xlabel("x");
ax[2].set_xlabel("x");
ax[0].set_ylabel("density");
ax[1].set_ylabel("density");
ax[2].set_ylabel("density");
plt.legend();
````

In this example, the transformation to the unconstrained space is $\log$ an its inverse is $\exp$, and we see the samples in the spaces are related by the transformation as expected:

````python
np.allclose(np.exp(unconstrained_samples), samples)
````

## 4. Multivariate Distributions

Multivariate distributions in PyMC do not define an inverse CDF, thus this class raises an exception if we try to use them directly:

````python
#### icdf NOT IMPLEMENTED
with pm.Model() as model:
  x = pm.MvNormal("x", mu=0.0, cov=np.eye(2))

try:
    PyMCWrapper(DigitalNetB2(2, seed=7), model)
except NotImplementedError as error:
    print(f"Caught Error: [\033[1m{str(error)[:49]}...]")
````

To work around this, we need to define our own multivariate distribution using i.i.d distributions. The model is defined with the i.i.d variables, which can be:
- sampled and transformed into the dependent joint distribution
- transformed **within the model** and used as an **input to other distributions**

````python
### equivalent 
def mv_normal_from_iid(x, mu, cov_chol):
    return mu + x @ cov_chol.T
    
with pm.Model() as model:
    cov = np.array([[1, 0.8],[0.8,1]])
    cov_chol = np.linalg.cholesky(cov)
    mu = np.array([-2, 2])
    x = pm.Normal("x", mu=0.0, sigma=np.ones_like(np.diag(cov)))
    ### we can also do distributions that depend on `x`.
    ### here we show a linear regression example
    # mu_y = mv_normal_from_iid(x, mu, cov_chol) @ beta 
    # y = pm.Normal("y" mu=mu_y, sigma=np.diag(2))

# take samples using MCMC
with model:
    idata = pm.sample(draws=1000, tune=1000)
mcmc_x_samples = np.asarray(idata.posterior["x"]).reshape(-1,2)
mcmc_x_samples = mv_normal_from_iid(mcmc_x_samples, mu, cov_chol)
````

````python
# take samples using QMC
mv_normal = PyMCWrapper(DigitalNetB2(2, seed=7), model)
qmc_samples = mv_normal(2**14)
qmc_samples = mv_normal_from_iid(qmc_samples, mu, cov_chol)
````

````python
plt.figure()
plt.scatter(qmc_samples[:,0], qmc_samples[:,1], alpha=0.1, label="qmc samples");
plt.scatter(mcmc_x_samples[:,0], mcmc_x_samples[:,1], label="mcmc samples", alpha=0.1)
plt.xlim(-5,1);
plt.ylim(-1,5);
plt.legend();
````

The QMC samples are visibly more representative of the distribution.

## 5. Sampling Prior Vs Posterior Models

PyMC provides utilities to sample the model **prior** (`sample_prior_predictive`), **posterior** (`sample`) and **posterior predictive** (`sample_posterior_predictive`) distributions. PyMCWrapper, on the other hand, **always samples the prior** distribution. Any observed variables are removed to avoid undefined behavior. Here we discuss how to sample from the posterior, posterior predictive and prior predictive distributions. The common thread is, further processing of the model must be performed. Such options include:
- Fitting an approximate distribution to the posteior (Variational Inference, i.e. `pymc.fit`), and sampling from that. 
- Condition Random variables of interest to point values such as posterior mode and sample the resulting model (`pymc.do`).

This is demonstrated below with more complicated model examples taken from [PosteriorDB](https://github.com/stan-dev/posteriordb/tree/master/posterior_database/models/pymc).

### Arma autoregression model

````python
# download data
import urllib.request, zipfile

url = "https://github.com/stan-dev/posteriordb/raw/refs/heads/master/posterior_database/data/data/arma.json.zip"
output_filename = "arma.json.zip"

# Downloads the file and saves it locally
urllib.request.urlretrieve(url, output_filename)
print(f"Downloaded {output_filename} successfully!")
zipfile.ZipFile(output_filename).extractall()  # overwrites existing files
````

````python
## https://github.com/stan-dev/posteriordb/blob/master/posterior_database/models/pymc/arma11.py
import pytensor
import pytensor.tensor as pt
import json
with open("arma.json", "r") as f:
    data_dict = json.load(f)
data_dict["y"] = np.asarray(data_dict["y"])
def arma(data, observe=True):
    y_obs = data["y"]  # Observed series
    up_to = data["T"]  # Timesteps
    with pm.Model() as pymc_model:
        mu = pm.Normal(
            "mu", mu=0, sigma=10
        )  # mean coefficient
        phi = pm.Normal(
            "phi", mu=0, sigma=2
        ) # autoregressive coef
        theta = pm.Normal(
            "theta", mu=0, sigma=2
        ) # moving average
        sigma = pm.HalfCauchy(
            "sigma", beta=2.5
        ) # noise scale

        # scan variables
        y = pt.as_tensor_variable(y_obs)
        y_hat0 = mu + phi * mu
        err0 = y[0] - y_hat0
        outputs_info = [y_hat0, err0, y[0]]

        def step(curr_obs, prev_pred, prev_err, prev_obs, mu, phi, theta):
            y_hat = mu + phi * prev_obs + theta * prev_err
            new_err = curr_obs - y_hat
            return [y_hat, new_err, curr_obs]

        [predictions, _, _] = pytensor.scan(fn=step,
                                              outputs_info=outputs_info,    
                                              sequences=y[1:],              
                                              non_sequences=[mu, phi, theta], return_updates=False)

        # concatenating predictions
        y_hat0_vec = y_hat0.dimshuffle("x")
        final_predictions = pt.concatenate([y_hat0_vec, predictions])
        if observe:
            pm.Normal("y", mu=final_predictions, sigma=sigma, observed=y)
        else:
            pm.Normal("y", mu=final_predictions, sigma=sigma)
    return pymc_model
_arma = arma(data_dict)
_arma
````

"Observed" random variables are treated as free variables.

#### Posterior Sampling

To sample from the posterior we need to estimate it, create a model for it, and then QMCPy can be used to sample that model. The simplest way to do this would be to first obtain an estimated mode for the model conditioned on the data. This point estimate model can be constructed with `pymc.do` and sampled:

````python
# PyMC MCMC sampling
# conditioned model:
pymc_point_estimate_arma = pm.do(arma(data_dict), {"mu": 0.1, "phi": 0.05, "theta": 0.1, "sigma":0.2})
with pymc_point_estimate_arma:
    # In the conditioned model, the prior predictive is equivalent to
    # the posterior predictive, because the "prior" here is now a point mass
    # that we set at the posterior mode
    idata_y = pm.sample_prior_predictive()
````

````python
# QMC sampling:
# conditioned model:
qmc_point_estimate_arma = pm.do(arma(data_dict, observe=False), {"mu": 0.1, "phi": 0.05, "theta": 0.1, "sigma":0.2})
qmc_point_estimate_arma = PyMCWrapper(DigitalNetB2(200, seed=7), qmc_point_estimate_arma)
qmc_point_samples = qmc_point_estimate_arma(2**13)
y_samples = qmc_point_samples
````

````python
import matplotlib
T = 200
plt.figure()
for i in range(3):
    plt.plot(np.asarray(idata_y.prior_predictive["y"]).reshape(-1,200)[i,:T], alpha=0.2, color="r", label="mcmc samples")
    plt.plot(y_samples[i*2,:T], alpha=0.2, color="g", label="qmc samples")
    
legend_handles = [
    # A custom solid line
    matplotlib.lines.Line2D([0], [0], color='red', label='mcmc samples'),
    # A dashed line with a marker
    matplotlib.lines.Line2D([0], [0], color='green', label='qmc samples')
]

# 2. Pass your custom handles directly into the legend
plt.legend(handles=legend_handles, loc='upper right');
````

Note how the conditioned model has no "free random variables" besides `y`.  
Accordingly the sampler dimensionality we used was 200 for the 200 `y` time steps.  
Also, note how the model PyMCWrapper wraps must use `observe=False` so that the variable `y` remains.  
This is contrasted with PyMC's `sample_prior_predictive` which assumes `y` is observed.  

````python
print(qmc_point_estimate_arma.model.free_RVs)
print(qmc_point_estimate_arma.model.observed_RVs)
````

````python
print(pymc_point_estimate_arma.free_RVs)
print(pymc_point_estimate_arma.observed_RVs)
````

### Variational Approximation sampling

A second way to sample from the posterior for this model is to fit a variational estimate of the posterior and sample that.  
This is currently not directly supported, but can be rolled out by representing the results of variational inference in a `Model` object. We demonstrate here for a simple regression model using the PyMC `fit` method. We need to convert the resulting `Approximation` object to a `Model` which `PyMCWrapper` can wrap and sample from:

````python
# define model and observations
x_obs = np.ones(10)*0.1
mu_sigma = 10.0
x_sigma = 0.3
with pm.Model() as observed_normal_model:
    mu = pm.Normal("mu", mu=0.0, sigma=mu_sigma)
    x = pm.Normal("x", mu=mu, sigma=x_sigma, observed=x_obs)
with observed_normal_model:
    normal_mcmc_idata = pm.sample()
with observed_normal_model:
    normal_mcmc_idata_posterior = pm.sample_posterior_predictive(normal_mcmc_idata)
with observed_normal_model:
    VI_normal = pm.ADVI().fit(obj_n_mc=1000)

########### extract VI Model
# Extract the unconstrained mean and standard deviation arrays
# (We use .eval() because these attributes are PyTensor symbolic variables)
flat_means = VI_normal.mean.eval()
flat_stds = VI_normal.std.eval()

# Map the flat arrays back to individual model variables using approx.ordering
fitted_gaussians = {}

for var_name, (_, slice_obj, _, _) in VI_normal.ordering.items():
    fitted_gaussians[var_name] = {
        "mean": flat_means[slice_obj],
        "std": flat_stds[slice_obj]
    }

with pm.Model() as VI_model:
    mu = pm.Normal("mu", mu=fitted_gaussians["mu"]["mean"], sigma=fitted_gaussians["mu"]["std"]) # VI posterior
    x = pm.Normal("x", mu=mu, sigma=0.3) # NO OBSERVATIONS!
````

Here we wrap the new model and compare the samples with the MCMC samples from the original model, including the posterior predictive samples.  
Note how the `x` and `mu` samples are not centered around the origin, as expected.

````python
qmc_posterior = PyMCWrapper(DigitalNetB2(2, seed=7), VI_model)
qmc_samples = qmc_posterior(2**13)
qmc_mu_samples = qmc_samples[:,0]
qmc_x_samples = qmc_samples[:,1:].ravel()

# analytic posterior
mus = np.linspace(-0.5,0.7,1000)
mu_post_sigma = np.sqrt(len(x_obs) / (x_sigma**2) + 1/mu_sigma**2) ** (-1)
mu_post_mean = mu_post_sigma ** 2 / (x_sigma**2) * np.sum(x_obs)
mu_post_pdf = stats.norm.pdf(mus, loc=mu_post_mean, scale=mu_post_sigma)

xs = np.linspace(-1.2,1.2,1000)
x_post_sigma = mu_post_sigma / np.sqrt(1-(1+mu_post_sigma**2/x_sigma**2)**(-1))
x_post_mean = mu_post_mean
x_post_pdf = stats.norm.pdf(xs, loc=x_post_mean, scale=x_post_sigma)

tonp = lambda x: np.asarray(x).ravel()
fig, ax = plt.subplots(1,2)
ax[0].set_title("mu")
ax[0].hist(qmc_mu_samples, bins=30, alpha=0.5, label="qmc", density=True);
ax[0].hist(tonp(normal_mcmc_idata.posterior["mu"]), bins=30, alpha=0.5, label="mcmc", density=True);
ax[0].plot(mus, mu_post_pdf, label="True PDF", color="black");
ax[1].set_title("x")
ax[1].hist(qmc_x_samples, bins=30, alpha=0.5, label="qmc", density=True);
ax[1].hist(tonp(normal_mcmc_idata_posterior.posterior_predictive["x"]), bins=30, alpha=0.5, label="mcmc", density=True);
ax[1].plot(xs, x_post_pdf, label="True PDF", color="black");
plt.legend();
````

## 6. Limitation - Model With Missing icdf

Here we demonstrate some of the limitations of `PyMCWrapper` that come from the incomplete implementation of `icdf` in PyMC itself.  
Generally as of PyMC `6.0.1` some discrete distributions and Dirichlet distribution, and possibly others, are missing such an implementation.

````python
observations = np.concat([1+np.random.randn(50)/10, -1+np.random.randn(50)/10])
n_components = 2

with pm.Model() as gauss_mix:
    mu = pm.Normal(
        "mu",
        mu=observations.mean(),
        sigma=10,
        shape=n_components,
        transform=pm.distributions.transforms.ordered
    )
    sigma = pm.HalfNormal("sigma", sigma=10, shape=n_components)
    weights = pm.Dirichlet("w", np.ones(n_components))

    y = pm.NormalMixture("y", w=weights, mu=mu, sigma=sigma, observed=observations)

with gauss_mix:
    idata = pm.sample(initvals={"mu": [-0.7, 0.7]})
    idata_predictive = pm.sample_posterior_predictive(idata)
````

````python
mu = np.asarray(idata.posterior["mu"]).reshape(-1, 2)
plt.figure()
plt.hist(mu[:,0], bins=100, label="component 0", color="red")
plt.hist(mu[:,1], bins=100, label="component 1", color="blue");
plt.legend();
plt.title("mu draws");
````

````python
w = np.asarray(idata.posterior["w"]).reshape(-1,2)
plt.figure()
plt.hist(w[:,0], bins=30, label="w component 0", density=True, alpha=0.3)
plt.hist(w[:,1], bins=30, label="w component 1", density=True, alpha=0.3)
plt.legend();
plt.title("w draws");
````

````python
y = np.asarray(idata_predictive.posterior_predictive["y"]).ravel()
plt.figure()
plt.hist(y, bins=100, density=True)
plt.title("y posterior predictive");
````
