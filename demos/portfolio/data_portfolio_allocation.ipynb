## 1. Download data from Yahoo Finance

```python
def download_data(tickers, description):
    """Batch-download adjusted close/volume for all tickers in one call."""
    col_names = ['Ticker', 'Company', 'Date', 'Adj Close Price', 'Volume']
    data = yf.download(tickers, start=start_date, end=end_date, group_by='ticker')
    rows = []
    for ticker, company in zip(tickers, description):
        sub = data[ticker].reset_index()
        price_col = 'Adj Close' if 'Adj Close' in sub.columns else 'Close'
        if sub[price_col].isna().all():
            print(f"Warning: no data for {ticker}")
            continue
        sub = sub[['Date', price_col, 'Volume']].copy()
        sub.insert(0, 'Company', company)
        sub.insert(0, 'Ticker', ticker)
        sub.columns = col_names
        sub['Adj Close Price'] = sub['Adj Close Price'].round(2)
        n_na = sub['Adj Close Price'].isna().sum()
        if n_na:
            print(f"Warning: {ticker} dropped {n_na} date(s) with missing price")
        rows.append(sub.dropna(subset=['Adj Close Price']))

    return pd.concat(rows, ignore_index=True)
```

```python
def get_log_ret(df):
    clean = bu.log_returns(df)
    n_dropped = df['Date'].nunique() - len(clean)
    print(f"{df['Ticker'].nunique()}-ticker log returns: {len(clean)} dates ({n_dropped} dropped for missing/NA values)")

    return clean
```

## 2. Risk-free rates

```python
def download_rf(start_date, end_date, ticker="^IRX"):
    """Download 13-week T-bill (^IRX) data and return the annual log risk-free rate."""
    df = yf.download(ticker, start=start_date, end=end_date).reset_index()

    rate_col = 'Adj Close' if 'Adj Close' in df.columns else 'Close'
    df = df[['Date', rate_col]].copy()
    df.rename(columns={rate_col: 'annual_rate_pct'}, inplace=True)

    df['rf_annual'] = np.log(1 + df['annual_rate_pct'] / 100)

    return df[['Date', 'rf_annual']].dropna().set_index('Date')['rf_annual']
```

## 3. S&P 500 benchmark

```python
def download_sp500(start_date, end_date, ticker="^SP500TR"):
    """Download S&P 500 total return (dividends reinvested), for use as a passive $ benchmark.

    Uses ^SP500TR (total return), not the price-only ^GSPC, because the
    individual stocks below use dividend-adjusted prices, a price-only
    benchmark would understate passive buy-and-hold by the dividend yield
    compounded over the period (confirmed empirically: the ^GSPC vs.
    ^SP500TR gap is ~24% over the full 2014-2025 window and ~6% over the
    out-of-sample years alone).

    Args:
        start_date (str): First date of the downloaded series.
        end_date (str): Last date of the downloaded series.
        ticker (str): Yahoo Finance ticker to download.

    Returns:
        pd.Series: Closing price indexed by Date.
    """

    df = yf.download(ticker, start=start_date, end=end_date)
    price_col = 'Adj Close' if 'Adj Close' in df.columns else 'Close'
    price = df[price_col]
    if isinstance(price, pd.DataFrame):
        price = price.iloc[:, 0]
    return price.rename('sp500_price').reset_index().dropna().set_index('Date')['sp500_price']
```

The two series are cached together in one file, `market_*.csv.gz`, with columns `sp500_price` and `rf_annual`, joined on `Date`; a date missing from one series is left empty in its column.

```python
def get_market(start_date, end_date, data_dir=data_dir):
    """Download and cache the S&P 500 and risk-free-rate series in one file."""
    p = Path(f"{data_dir}market_{start_date}_to_{end_date}.csv.gz")
    if p.exists():
        return pd.read_csv(p, index_col='Date', parse_dates=True)
    market = pd.concat([download_sp500(start_date, end_date), download_rf(start_date, end_date)], axis=1).sort_index()
    market.to_csv(p, compression={'method': 'gzip', 'mtime': 0})
    return market


market = get_market(start_date, end_date)
```

```python
def get_ticker_data(tickers, descriptions, data_dir=data_dir, start_date=start_date, end_date=end_date):
    """Download and cache the canonical ticker data and its log returns."""
    price_path = Path(f"{data_dir}df40_{start_date}_to_{end_date}.csv.gz")
    lr_path = Path(f"{data_dir}log_returns40_{start_date}_to_{end_date}.csv.gz")

    if price_path.exists():
        df = pd.read_csv(price_path, parse_dates=['Date'])
    else:
        df = download_data(tickers, descriptions)
        df.to_csv(price_path, index=False, compression={'method': 'gzip', 'mtime': 0})

    lr = get_log_ret(df)
    lr.to_csv(lr_path, compression={'method': 'gzip', 'mtime': 0})
    return df, lr
```

## 5. Checksums (for a published data snapshot)

Computed so a frozen copy of this data directory published elsewhere (e.g. a GitHub release asset) can be verified against what this notebook actually produced, independent of whether that copy sits in this repo. Only the downloaded files are listed; the nested asset universes and their log returns are derived from the canonical 40-asset price file.

```python
import hashlib

checksums_path = Path(f"{data_dir}checksums.txt")
with open(checksums_path, "w") as f:
    for p in sorted(Path(data_dir).glob("*.csv.gz")):
        if p.name.startswith("log_returns"):
            continue
        digest = hashlib.sha256(p.read_bytes()).hexdigest()
        f.write(f"{digest}  {p.name}\n")
print(checksums_path.read_text())
```
